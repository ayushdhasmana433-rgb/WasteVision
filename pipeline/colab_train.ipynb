```python
import os, tensorflow as tf
gpus = tf.config.list_physical_devices("GPU")
msg = [f"TensorFlow {tf.__version__}", f"GPU: {gpus}"]
for p in ["/content/drive/MyDrive/WasteVision_v3/data/splits_v3.csv",
          "/content/drive/MyDrive/WasteVision_final/data/normalized_images.tar",
          "/content/drive/MyDrive/WasteVision_final/data/all_images_before_dedup.csv",
          "/content/drive/MyDrive/WasteVision_v3/data/extra_images.tar"]:
    msg.append(("OK      " if os.path.exists(p) else "MISSING ") + p)
print("\n".join(msg)); open("/content/drive/MyDrive/WasteVision_v3/logs/train_check.log", "a").write("\n".join(msg) + "\n")
assert gpus, "No GPU: Runtime > Change runtime type > T4 GPU, then Run all again"
assert not any(m.startswith("MISSING") for m in msg), "input file missing"
print("CHECK OK")
```
